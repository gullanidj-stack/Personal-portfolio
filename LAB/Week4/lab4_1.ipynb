#**● Lab 4.1: Student Management System (List Operations)**

○ Objective: Students will build a simple program to manage a list of student names,
demonstrating various list operations.

○ Tasks (Common for Local &amp; Colab):

1. Create a new Python file (lab4_1.py) or Colab code cell.

2. Initialize an empty list called student_names.

3. Add Students: Prompt the user to enter 3-5 student names and append() each
name to the list. Print the current list.

4. Find a Student: Ask the user for a name to search for. Use if item in list: to check
if the student is present. If found, print their index using index().

5. Remove a Student: Ask the user for a student name to remove. Use remove() to
delete the student. Handle the case where the student is not found (e.g., using if
item in list: before remove()). Print the updated list.

6. Sort Students: Sort the student_names list alphabetically using sort(). Print the
sorted list.

7. Count Students: Print the total number of students in the list using len().

8. Challenge: Implement a simple menu-driven system using a while loop (from
Week 3) where the user can choose to add, remove, search, sort, or exit.

```python
print("--- Lab 4.1: Student Management System ---")

student_names = []

print("\n[1] Add Students")
for i in range(5):
    name = input(f"Enter student name #{i + 1}: ")
    student_names.append(name)

print(f"Current list: {student_names}")

print("\n[2] Find a Student")
search_name = input("Enter a name to search for: ")

if search_name in student_names:
    position = student_names.index(search_name)
    print(f"Found '{search_name}' at index {position}")
else:
    print(f"'{search_name}' not found in the list")

print("\n[3] Remove a Student")
remove_name = input("Enter a name to remove: ")

if remove_name in student_names:
    student_names.remove(remove_name)
    print(f"'{remove_name}' has been removed")
else:
    print(f"'{remove_name}' not found, nothing removed")

print(f"Updated list: {student_names}")

print("\n[4] Sort Students")
student_names.sort()
print(f"Sorted list: {student_names}")


print("\n[5] Count Students")
print(f"Total students: {len(student_names)}")
```

#**Challenge:**

```python
print("--- Lab 4.1: Student Management System ---")
student_names = []

while True:
  print("=Menu=")
  print("[1] Add Students")
  print("[2] Find a Student")
  print("[3] Remove a Student")
  print("[4] Sort Students")
  print("[5] Count Students")
  print("[6] Exit")

  choice = input("Choose a menu (1-6) : ")

  if choice == "1" :
    for i in range(5):
       name = input(f"Enter student name #{i + 1}: ")
       student_names.append(name)
       print(f"Current list: {student_names}")

  elif choice == "2" :
    search_name = input("Enter a name to search for: ")
    if search_name in student_names:
      position = student_names.index(search_name)
      print(f"Found '{search_name}' at index {position}")
    else:
      print(f"'{search_name}' not found in the list")

  elif choice == "3" :
    remove_name = input("Enter a name to remove: ")
    if remove_name in student_names:
      student_names.remove(remove_name)
      print(f"'{remove_name}' has been removed")
    else:
      print(f"'{remove_name}' not found, nothing removed")
    print(f"Updated list: {student_names}")

  elif choice == "4" :
    student_names.sort()
    print(f"Sorted list: {student_names}")

  elif choice == "5" :
    print(f"Total students: {len(student_names)}")

  elif choice == "6" :
    print("Thank you!")
    break

  else:
    print("Invalid choice. Please try again.")




```
