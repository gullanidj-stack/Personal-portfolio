#**● Lab 4.2: Geometric Calculations (Tuples &amp; Immutability)**

○ Objective: Students will use tuples to represent 2D points and calculate the distance

between them, highlighting tuple immutability.

○ Tasks (Common for Local &amp; Colab):

1. Create a new Python file (lab4_2.py) or Colab code cell.

2. Define Points: Define two points as tuples: point1 = (3, 4) and point2 = (6, 8).

3. Access Coordinates: Access and print the x and y coordinates of both points.
■ Example: &quot;Point 1: x = 3, y = 4&quot;

4. Attempt Modification (Illustrate Immutability): Try to change point1&#39;s x-
coordinate (e.g., point1[0] = 5). Observe and explain the TypeError. (Comment out
this line after demonstrating).

5. Calculate Distance:

■ Import the math module: import math.

■ Use the distance formula: (x2​−x1​)2+(y2​−y1​)2​.

■ Calculate the distance between point1 and point2 using math.sqrt() and **
(power operator).

■ Print the calculated distance.

6. Challenge: Prompt the user to enter coordinates for two new points and then
calculate the distance between them. Ensure type conversion from input().

```python
import math

print("--- Lab 4.2: Geometric Calculations ---")

point1 = (3,4)
point2 = (6,8)

print("\n[1] Access Coordinates")
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

print("\n[2] Attempt Modification (Immutability Demonstration)")
try:
    point1[0] = 5
except TypeError as e:
    print(f"Error trying to modify tuple: {e}")

print("\n[3] Calculate Distance")
distance = math.sqrt((point2[0] - point1[0]) ** 2 + (point2[1] - point1[1]) ** 2)
print(f"Distance between point1 {point1} and point2 {point2}: {distance}")
```

#**Activity 4.2: Interactive To-Do List (using List Operations)**

สิ่งที่อยากให้ทำ : ให้นักเรียนพัฒนาโปรแกรมจัดการรายการสิ่งที่ต้องทำ (To-Do List) แบบข้อความ (Text-based) อย่างง่าย โดยประยุกต์ใช้ความรู้เรื่องการจัดการ List ที่ได้เรียนมา

สิ่งที่ต้องมี

1. สร้าง List ว่างเริ่มต้นไว้ตัวหนึ่ง ชื่อว่า todo_list

2. ใช้การวนลูป while เพื่อแสดงเมนูคำสั่งให้ผู้ใช้เลือก ดังนี้:

    2.1 เพิ่มรายการงาน (Add a task)

    2.2 ดูรายการงานทั้งหมด (View tasks)

    2.3 ทำเครื่องหมายว่างานเสร็จสิ้นแล้ว (ลบงานออก โดยเลือกจากตำแหน่ง Index หรือตามชื่อคำสั่ง)

    2.4 ออกจากโปรแกรม (Exit)

3. ใช้คำสั่ง input() เพื่อรับค่าเมนูที่ผู้ใช้เลือก

4. ใช้คำสั่งจัดการ List ต่างๆ (เช่น append(), remove() หรือ pop(), len()) ในการบริหารจัดการข้อมูลใน todo_list

```python
todo_list = []

while True :
  print("\n= To-Do List =")
  print("[1] Add a task")
  print("[2] View tasks")
  print("[3] Complete a task")
  print("[4] Exit")

  choice = input("Choose a menu (1-4) : ")

  if choice == "1" :
    task =  input(f"Enter a new task : ")
    todo_list.append(task)
    print("Task added successfully!")

  elif choice == "2" :
    if len(todo_list) == 0 :
      print(f"No task in To-Do List.")
    else:
      print("\nYour To-Do List:")
      for i in range(len(todo_list)):
        print(f"{i+1}. {todo_list[i]}")

  elif choice == "3":
    if len(todo_list) == 0:
        print("No tasks to remove.")
    else:
        print("\nYour To-Do List:")
        for i in range(len(todo_list)):
            print(f"{i+1}. {todo_list[i]}")

        index = int(input("Enter the task number to remove: "))

        if 1 <= index <= len(todo_list):
            completed = todo_list.pop(index - 1)
            print(f"Task '{completed}' completed and removed!")
        else:
            print("Invalid number.")

  elif choice == "4" :
    print("Thank you!")
    break

  else:
    print("Invalid choice. Please try again.")
```
