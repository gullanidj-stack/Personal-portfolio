Countdown & User Guessing Game
Objective:

  Students will implement while loops for a countdown and a simple guessing game, incorporating break for early exit.

Part 1: Countdown Timer:

Prompt the user to enter a starting number for the countdown.
Use a while loop to count down from that number to 0, printing each number.
After the loop, print "Blast off!"



```python
start = int(input("Enter starting number for countdown : "))
while start >=0:
  print(f"{start}")
  start -= 1
print("Blast off!")
```

Part 2: Simple Guessing Game:

Choose a secret number (e.g., secret_number = 42).

Use a while True loop to continuously ask the user to guess the number.

If the guess is too low, print "Too low! Try again."

If the guess is too high, print "Too high! Try again."

If the guess is correct, print "Congratulations! You guessed it!" and use break to exit the loop.



```python
secret_number = 99

while True:
  guess = int(input("Guess the number : "))
  if guess < secret_number:
    print("Too low! Try again.")
  elif guess > secret_number:
    print("Too high! Try again.")
  else:
    print("Congratulations! You guessed it!")
    break
```

Challenge: Add a limited number of attempts (e.g., 5 attempts) and use a break or else clause for the loop if attempts run out.

```python
secret_number = 99
attempts = 5

while attempts > 0:
  guess = int(input("Guess the number : "))
  if guess < secret_number:
    print("Too low! Try again.")
  elif guess > secret_number:
    print("Too high! Try again.")
  else:
    print("Congratulations! You guessed it!")
    break

  attempts -= 1
  print("Attempts left: ", attempts)

if attempts == 0:
  print("Game Over!")
```

Activity 3.2: Pattern Printing with Nested Loops

Format: Individual creative coding challenge.

A square of asterisks:

```python
for i in range(5):
  for j in range(5):
    print("*", end=" ")
  print()
```

A right-angled triangle of asterisks:

```python
for i in range(1,6):
  for j in range(i):
    print("*", end=" ")
  print()
```

Challenge: An inverted right-angled triangle or a pyramid.
Purpose: Reinforce understanding of nested loops and the relationship between loop variables and printed output.
Encourages algorithmic thinking and problem-solving.


```python
for i in range(5):
  for j in range(5-i):
    print("*", end=" ")
  print()
```
