```python
# Puzzler 1
x = 10
y = 3.5
z = x + y
print(z)
print(type(z))
```

ได้ผลลัพธ์: 13.5

ประเภทDataType: float

ทำไมถึงเป็นtype float:y เป็นข้อมูลtype floatอยู่แล้ว เพราะฉะนั้นเมื่อบวกกันกับxที่เป็น int Python จะแปลงค่าเป็นfloatให้โดยอัตโนมัติ

```python
# Puzzler 2
name = "Alice"
age = "30" # Note: This is a string
message = name + " is " + age + " years old."
print(message)
print(type(age))
```

ได้ผลลัพธ์: Alice is 30 years old.

ประเภทDataType: String

ทำไมถึงเป็นtype String: เพราะ age ถูกกำหนดค่าใน" " ซึ่งPython จะมองเป็นข้อความ ไม่ใช่ตัวเลข

```python
# Puzzler 3
is_sunny = True
is_raining = False
print(is_sunny and is_raining)
print(type(is_sunny))
```

ได้ผลลัพธ์: False

ประเภทDataType: Boolean

ทำไมถึงเป็นtype Boolean: is_sunny เก็บค่าtrue ซึ่งเป็นข้อมูลtype Boolean อยู่แล้ว

```python
# Puzzler 4
num_str = input("Enter a number: ")
result = int(num_str) + 5
print(result)
print(type(result))
```
