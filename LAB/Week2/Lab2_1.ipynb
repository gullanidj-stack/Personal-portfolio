● Lab 2.1: Number Classifier
○ Objective: Students will write a script to classify a user-provided integer as positive,
negative, or zero, and also check if it&#39;s even or odd.
○ Tasks (Common for Local &amp; Colab):
1. Create a new Python file (lab2_1.py) or Colab code cell.
2. Prompt the user to enter an integer using input().
3. Convert the input to an integer.
4. Use if-elif-else to determine if the number is positive, negative, or zero. Print the
classification.
5. Use an if-else statement to determine if the number is even or odd. Print the result.
6. Challenge: Combine the checks using logical operators. For example, print &quot;The
number is positive and even.&quot; or &quot;The number is negative and odd.&quot;

```python
num = input("Enter a Number : ")
num = int(num)
if num > 0:
  intnum = "The number is positive"
elif num < 0:
  intnum = "The number is negative"
else:
  intnum = "The number is zero"

if num % 2 == 0:
  parity = "even."
else:
  parity = "odd."

print(intnum + " and " + parity)


```

Block A: Evaluation Check python

```python
temp = 25
is_raining = True
if temp > 20 and not is_raining:
    print("Perfect day for a picnic!")
elif temp <= 20 or is_raining:
    print("Stay indoors or bring an umbrella.")
else:
    print("Unsure about the weather.")
```

ลำดับการทำงานทีละขั้น:

1. temp > 20 → 25 > 20 ผลคือ True เพราะ อุณหภูมิมากกว่า 20 จริง
2. not is_raining → not True ผลคือ False เพราะฝนตกอยู่ (is_raining = True)
3. True and False ผลคือ False เพราะ เงื่อนไข if เป็นเท็จ → ข้ามไป elif
4. temp <= 20 → 25 <= 20 ผลคือ False
5. is_raining ผลคือ True
6. False or True ผลคือ True เพราะ เงื่อนไข elif เป็นจริง → เข้าบล็อกนี้

Output:
Stay indoors or bring an umbrella.

Block B: Vectorized Filter

```python
phrase = "Python 2026 Engine"
consonants = {c.lower() for c in phrase if c.isalpha() and c.lower() not in 'aeiou'}
print(f"Unique consonants: {consonants}")
```

ลำดับการทำงานทีละขั้น:

1. วนลูปทีละตัวอักษรใน "Python 2026 Engine"
2. เงื่อนไข c.isalpha() → คัดเฉพาะตัวอักษร (ตัด 2, 0, 2, 6, และช่องว่างทิ้ง)
3. เงื่อนไข c.lower() not in 'aeiou' → คัดเฉพาะพยัญชนะ (ตัดสระ a, e, i, o, u ทิ้ง)
4. c.lower() → แปลงเป็นตัวพิมพ์เล็กทั้งหมด
5. เก็บใน set ({}) → set จะไม่เก็บค่าซ้ำโดยอัตโนมัติ
6. ตัวอักษรที่ผ่านการกรอง: p, y, t, h, n, n, g, n → เหลือไม่ซ้ำ: p, y, t, h, n, g

Output: Unique consonants: {'h', 'y', 'p', 'g', 't', 'n'}
