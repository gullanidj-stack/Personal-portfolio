Lab 2.2: Simple Decision-Making Program (Movie Rating)
○ Objective: Students will create a program that recommends a movie rating based on
the user&#39;s age, demonstrating nested or chained conditionals.
○ Tasks (Common for Local &amp; Colab):
1. Create a new Python file (lab2_2.py) or Colab code cell.
2. Prompt the user to enter their age. Convert it to an integer.
3. Implement the following logic:
 If age is less than 5: Print &quot;You&#39;re too young for movies! Enjoy cartoons.&quot;
 If age is 5 to 12 (inclusive): Recommend &quot;G-rated or PG-rated movies.&quot;
 If age is 13 to 17 (inclusive): Recommend &quot;PG-13 or R-rated (with parental
guidance).&quot;
 If age is 18 or older: Recommend &quot;Any movie rating.&quot;
4. Challenge: Add a secondary prompt: &quot;Do you like action movies? (yes/no)&quot;. If the
age is 18 or older AND they like action, print &quot;You might enjoy the latest action
blockbuster!&quot; (This requires a nested if or more complex logical and conditions).

```python
age = input("Enter your age: ")
age = int(age)

if age < 5:
  print("You're too young for movies! Enjoy cartoons!")
elif age < 13:
  print("Recommended: G-rated or PG-rated movies.")
elif age < 18:
  print("Recommended: PG-13 or R-rated (with parental guidance).")
else:
  print("Recommended: Any movie rating.")
  like_action= input("Do you like action movies? (Yes/No) : ")
  if age >= 18 and like_action.lower() == "yes":
    print("You might enjoy the latest action blockbuster!")
  else:
    print("Recommended: Other genres suited to your age group")




```

● Activity 2.2: Design a Simple &quot;Choose Your Own Adventure&quot; Text Game

○ Instructions:
1. Brainstorm: As a group, come up with a very short &quot;choose your own adventure&quot;

2. Design: Outline 2-3 decision points in your story. For each decision, determine 2-3
possible choices and their outcomes.
3. Pseudo-code/Outline: Write down the basic input() prompts and the if-elif-else
structure needed to implement your story&#39;s decisions. You don&#39;t need to write full

4. Share: Each group presents their story concept and their conditional logic outline
to the class.

○ Purpose: Apply conditional logic to a creative, problem-solving scenario, fostering
design thinking and collaboration before coding.
Sample Snippets for Class and Colab Ready

```python
print("Welcome to the Three Little Pigs game!")

material = input("A big bad wolf is coming to blow down your house! \nChoose your building material 1=Straw 2=Wood 3=Brick: ")
material = int(material)

if material == 1:
    # Straw
    print("\nThe straw house was built fast, but it looks flimsy")
    print("The wolf huffs just once and the house blows away instantly!")
    action = input("Do you repair the house or run away? (repair/run): ")

    if action == "repair":
        print("\nYou rush to repair the house, but the wolf huffs again!")
        print("The house finally collapses. You die!")
    elif action == "run":
        print("\nYou decide to run just in time!")

        house = input("Which house are you running to (2=Wood house/3=Brick house): ")
        if house == "2":
            print("The wooden house took a while to fall, but it still collapsed. You die!")
        elif house == "3":
            print("You escape to your youngest sibling's brick house - You survive!")
        else:
            print("Invalid answer. There's no other house, you die!")
    else:
        print("Invalid answer. Please type 'repair' or 'run' only.")

elif material == 2:
    # Wood
    print("\nThe wooden house took a while, and looks fairly sturdy.")
    print("The wolf huffs harder! The house shakes but doesn't collapse yet.")
    action = input("Do you repair the house or run away? (repair/run): ")

    if action == "repair":
        print("\nYou rush to repair the house, but the wolf huffs again!")
        print("The house finally collapses. You die!")
    elif action == "run":
        print("\nYou decide to run just in time!")

        house = input("Which house are you running to (1=Straw house/3=Brick house): ")
        if house == "1":
            print("The straw house is already destroyed, there's nowhere to hide. You die!")
        elif house == "3":
            print("You escape to your youngest sibling's brick house - You survive!")
        else:
            print("Invalid answer. There's no other house, you die!")
    else:
        print("Invalid answer. Please type 'repair' or 'run' only.")

elif material == 3:
    # Brick
    print("\nThe brick house took the longest to build, but looks very strong.")
    print("No matter how hard the wolf huffs, the house doesn't budge!")
    print("The wolf gives up, exhausted.")
    print("Congratulations! The brick house is the strongest. You survive!")

else:
    print("Please choose only 1, 2, or 3.")
```
